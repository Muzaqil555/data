# Proje Başlığı: King County House Sales Prediction


```python
import pandas as pd

# Dataset'i oxu
df = pd.read_csv("kc_house_data.csv")

# Hər sütunun data type-larını göstər
df.dtypes
```

```python
df.drop(["id",], axis=1, inplace=True)
df.describe()
```

```python
df['floors'].value_counts().to_frame()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x="waterfront", y="price", data=df)
plt.show()
```

```python
sns.regplot(x="sqft_above", y="price", data=df)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = df[['sqft_living']]
y = df['price']

lm = LinearRegression()
lm.fit(X, y)

print("R^2:", lm.score(X, y))
```

```python
features = ["floors", "waterfront", "lat", "bedrooms",
            "sqft_basement", "view", "bathrooms",
            "sqft_living15", "sqft_above", "grade", "sqft_living"]

X = df[features]
y = df['price']

lm = LinearRegression()
lm.fit(X, y)

print("R^2:", lm.score(X, y))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures

Input = [('scale', StandardScaler()),
         ('polynomial', PolynomialFeatures(degree=2, include_bias=False)),
         ('model', LinearRegression())]

pipe = Pipeline(Input)

pipe.fit(X, y)
print("R^2:", pipe.score(X, y))
```

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=1)

RidgeModel = Ridge(alpha=0.1)
RidgeModel.fit(X_train, y_train)

print("R^2 test:", RidgeModel.score(X_test, y_test))
```

```python
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

RidgeModel2 = Ridge(alpha=0.1)
RidgeModel2.fit(X_train_poly, y_train)

print("R^2 test:", RidgeModel2.score(X_test_poly, y_test))
```
